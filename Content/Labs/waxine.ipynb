# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

Verklaring 1 lijkt onwaarschijnlijk. De zuurstof wordt wel 'gebruikt' maar wordt omgezet in CO_2 gas, dus heeft dit weinig invloed op de druk.

Verklaring 2 klopt maar is niet volledig compleet. Omdat de lucht uitzet ontsnapt er een deel via het water uit de buis. Als de lucht dan afkoelt is er in totaal minder lucht in de buis waardoor er minder druk is dan buiten. De druk duwt het water omhoog tot deze druk overeenkomt. Er waren ook bubbels, wat

Verklaring 3 lijkt ook erg onwaarschijnlijk. Bij het verbranden ontstaat er namelijk geen waterdamp vanuit de vlam. De damp in de lucht blijft daar, het enige waterdamp wat ontstaat komt door de hitte van de vlam op het vloeibare water. Deze kan je echter niet meetellen want als dat condenseert ben je terug waar je was, met dat water als vloeistof. Dit effect is niet verantwoordelijk het stijgende water.

Observaties:
Een waxinelichtje is in een laagje water geplaatst, aangestoken en een glazen buis is er overeen geplaatst.
150 ml water gebruikt.
Water in beker komt op start op 455-452=3ml hoog
Water is na kaars 455-399=56ml ongeveer 3cm verschil
er zijn bubbels geobserveerd, voordat de kaars stijgt.
